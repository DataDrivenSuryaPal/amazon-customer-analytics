## Calculation Validation

Existing numerical calculations in the dataset are validated to ensure that derived values are mathematically consistent with their underlying fields.

Numerical comparisons are performed using NumPy's `isclose()` function where appropriate.

This helps identify discrepancies caused by:

* Incorrect calculations
* Rounding differences
* Data-entry issues
* Inconsistent values

Only after validation is the dataset used for further feature engineering.

Folmula:

Quantity * UnitPrice *(1-Discount) + tax + ShippingCost = TotalAmount

```python
diff = (df_copy["Quantity"]*df_copy["UnitPrice"]*(1-df_copy["Discount"])+df_copy["Tax"]+df_copy["ShippingCost"])-df_copy["TotalAmount"]
mask = (~np.isclose(diff,0,atol=0.01))
print(mask.sum())
```

```python
df_copy.loc[mask,["Quantity","UnitPrice","Discount","Tax","ShippingCost","TotalAmount"]].head(10)
```

## Feature Engineering

New analytical features are created from the existing dataset to make the data more useful for business analysis.

### GrossAmount

Represents the order value before applying discounts.

### DiscountedAmount

Represents the discount-related amount calculated from the available pricing and discount information.

### NetAmount

Represents the final order value after considering the applicable discount.

### OrderYear

Extracted from `OrderDate` to support yearly analysis.

### OrderMonth

Extracted from `OrderDate` to support monthly analysis.

These engineered features will make it easier to answer business questions related to sales, discounts, customer behavior, and order trends.

```python
#Creating Gross Amount
df_copy["GrossAmount"] = (df_copy["Quantity"]*df_copy["UnitPrice"]).round(2)
df_copy["GrossAmount"].head(5)
```

```python
#Creating Discounted Amount
df_copy["DiscountedAmount"] = (df_copy["GrossAmount"]*df_copy["Discount"]).round(2)
df_copy["DiscountedAmount"].head(10)
```

```python
#Creating Net Amount
df_copy["NetAmount"] = (df_copy["GrossAmount"]-df_copy["DiscountedAmount"])
df_copy["NetAmount"]
```

```python
##Creating Order Year
df_copy["OrderYear"] = df_copy["OrderDate"].dt.year
df_copy["OrderYear"]
```

```python
##Creating Order Month
df_copy["OrderMonth"] = df_copy["OrderDate"].dt.month_name()
df_copy["OrderMonth"]
```

```python
df_copy.head(2)
```

## Reorganize the columns

```python
cols = df_copy.columns.tolist()
for c in ["OrderYear","OrderMonth","GrossAmount","DiscountedAmount","NetAmount"]:
    cols.remove(c)
idx = cols.index("OrderDate")+1
cols[idx:idx] = ["OrderYear","OrderMonth"]

idx = cols.index("TotalAmount")
cols[idx:idx] = ["GrossAmount","DiscountedAmount","NetAmount"]
df_copy = df_copy[cols]
```
