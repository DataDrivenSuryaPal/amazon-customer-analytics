## Total number of orders

```python
total_orders = df_copy["OrderID"].count()
print("Total number of orders =",total_orders)
```

## Total number of customers

```python
total_customers = df_copy["CustomerID"].count()
print("Total number of customers =",total_customers)
```

## Total number of products

```python
total_products = df_copy["ProductID"].count()
print("Total number of products =",total_products)
```

## Total number of seller

```python
total_seller = df_copy["SellerID"].count()
print("Total number of seller =",total_seller)
```

## Total number of brands

```python
total_brands = df_copy["Brand"].count()
print("Total number of brands =",total_brands)
```

## Total number of categories

```python
total_category = df_copy["Category"].count()
print("Total number of categories =",total_category)
```

## Date range

##First change the orderdate column to datetime

```python
df_copy["OrderDate"]=pd.to_datetime(df_copy["OrderDate"])
```

##Now find the daterange

```python
print ("Order date range",df_copy["OrderDate"].min(),
("To"),
df_copy["OrderDate"].max())
```

## Total number of countries and also the names of countries

##Total number of countries

```python
total_countries = df_copy["Country"].nunique()
print("Number of countries",total_countries)
```

##Name of the countries 

```python
country_names = df_copy["Country"].unique()
print("Name of countries",country_names)
```

## Percentage of orders delivered

```python
delivered_percentage = (df_copy["OrderStatus"].eq("Delivered").sum()/len(df_copy))*100
print("Delivery percentage =",f"{delivered_percentage:.2f}%")
```

## Percentage are cancelled

```python
cancelled_percentage = (df_copy["OrderStatus"].eq("Cancelled").sum()/len(df_copy))*100
print("Cancelled percentage =",f"{cancelled_percentage:.2f}%")
```
